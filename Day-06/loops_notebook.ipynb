

## What is a Loop?

A loop is used to repeat a task multiple times.

Instead of writing the same code again and again, a loop allows us to run it automatically.

---

# For Loop

## What is a For Loop?

A **for loop** is used when we know how many times we want to repeat something.

It runs a fixed number of times.

## Simple Idea

Repeat something for a specific number of times.

## Real-World Examples

* Taking 5 steps → you already know the count (5)
* Doing 10 push-ups → fixed number
* Calling 10 students one by one
* Printing numbers from 1 to 10

👉 In all these cases, the number of repetitions is already known.

---

# While Loop

## What is a While Loop?

A **while loop** is used when we do not know how many times the task will repeat.

It runs until a condition becomes false.

## Simple Idea

Repeat something until a condition is satisfied.

## Real-World Examples

* Keep walking until you reach home
* Keep studying until exam time
* Keep entering password until it is correct
* Keep filling water until the tank is full

👉 In all these cases, we don’t know exact repetitions — we only know the condition.

---

# Difference Between For Loop and While Loop

| For Loop                            | While Loop                                    |
| ----------------------------------- | --------------------------------------------- |
| Used when repetition count is known | Used when repetition depends on condition     |
| Fixed number of times               | Unknown number of times                       |
| Easier for beginners                | More flexible but needs careful handling      |
| Less chance of error                | Can cause infinite loop if condition is wrong |

---

# Easy Understanding

* **For Loop = Count based repetition**
* **While Loop = Condition based repetition**

---

# One-Line Concept

* For loop runs **fixed times**
* While loop runs **until condition is true**


# FOR LOOP EXAMPLES

## Example 1: Print Hello

```python
for i in range(5):
    print('Hello')
```

## Example 2: Numbers 1 to 5

```python
for i in range(1,6):
    print(i)
```

## Example 3: Even numbers

```python
for i in range(1,11):
    if i%2==0:
        print(i)
```

## Example 4: Reverse

```python
for i in range(5,0,-1):
    print(i)
```

## Example 5: List loop

```python
marks=[80,70,90]
for m in marks:
    print(m)
```

## Example 6: Sum

```python
total=0
for i in range(1,6):
    total+=i
print(total)
```

## Example 7: Nested loop

```python
for i in range(3):
    for j in range(3):
        print('*',end=' ')
    print()
```

## Example 8: String loop

```python
for ch in 'Python':
    print(ch)
```

## Example 9: Range step

```python
for i in range(0,10,2):
    print(i)
```

# WHILE LOOP EXAMPLES

## Example 1

```python
i=1
while i<=5:
    print(i)
    i+=1
```

## Example 2 Reverse

```python
i=5
while i>=1:
    print(i)
    i-=1
```

## Example 3 Sum

```python
i=1
total=0
while i<=5:
    total+=i
    i+=1
print(total)
```

## Example 4 Password

```python
password=''
while password!='1234':
    password=input('Enter password: ')
print('Access Granted')
```

## Example 5 Break

```python
i=1
while i<=10:
    if i==5:
        break
    print(i)
    i+=1
```
