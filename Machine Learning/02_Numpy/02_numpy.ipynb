```python
a = np.arange(8)
print(a)
```

# How to convert a 1D array into a 2D array (how to add a new axis to an array)

```python
a = np.array([1, 2, 3, 4, 5, 6])
print(a)
a.shape
```

```python
a2 = a[np.newaxis, :]
a2.shape
```

```python
row_vector = a[: ,np.newaxis]
row_vector.shape
```

```python
z = np.array([1, 2, 3, 4, 5, 6])
z.shape
```

```python
b = np.expand_dims(z, axis=1)
b.shape
```
