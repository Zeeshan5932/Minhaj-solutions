```python
data = {
    " Student Name ": [" Ali ", "Sara", "Ahmed", "Ali ", "Bilal", "Hina", None, "Usman", "Maham", "Ahmed", "Zara", "Hamza", "Ayesha", "sara", "Danish"],
    "Age": [20, 22, None, 20, -5, 21, 23, 24, None, 19, 22, 200, 18, 22, 25],
    "City ": ["lahore", "Lahore", "LHR", "lahore", "karachi", None, "Islamabad", "islamabad ", "KHI", "LHR", "Karachi", "lahore", "islamabad", "Lahore", "Multan"],
    "Gender": ["M", "F", "M", "M", "M", "F", "F", "Male", "Female", "M", "F", "M", "F", "F", None],
    "Marks": [85, 90, None, 85, 70, 88, 92, -10, 78, 81, None, 95, 76, 90, 89],
    "Admission Date": ["2024-01-10", "2024/01/12", "10-01-2024", "2024-01-10", None, "2024-02-01", "2024-02-05", "2024-02-08", "2024-03-01", "2024-03-05", "2024-03-10", "invalid-date", "2024-04-01", "2024-01-12", "2024-04-15"],
    "Unnamed: 0": list(range(1, 16))
}

messy_df = pd.DataFrame(data)
messy_df
```

## 3. Save and Load Dataset

Real case mein hum CSV file read karte hain. Is liye pehle dataset ko CSV mein save karen gy, phir read karen gy.

```python
# messy_df.to_csv("messy_student_dataset.csv", index=False)

df = pd.read_csv("./data/messy_student_dataset.csv")
df.head()
```

## 4. Check Dataset Shape

`shape` se rows aur columns ka count pata chalta hai.

```python
df.shape
```

## 5. Check Column Names

Column names mein extra spaces ya unwanted columns ho sakte hain.

```python
df.columns
```

## 6. Check Basic Information

`info()` se non-null values aur data types pata chalte hain.

```python
df.info()
```

## 7. Summary Statistics

`describe()` numeric columns ka basic summary deta hai.

```python
df.describe()
```

## 8. Check Missing Values

Missing values ka count har column ke liye check karte hain.

```python
df.isnull().sum().sort_values(ascending=True)
```

## 9. Check Duplicate Rows

Duplicate rows analysis ko wrong kar sakti hain.

```python
df.duplicated().sum()
```

## 10. Clean Column Names

Column names ko simple aur readable banate hain:
- extra spaces remove
- lowercase
- spaces ko underscore mein convert

```python
df.columns = df.columns.str.strip()
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_")

df.columns
```

## 11. Drop Unnecessary Column

`unnamed:_0` jaisi columns aksar index ki wajah se ban jati hain. Isko remove kar dete hain.

```python
df = df.drop(columns=["Unnamed: 0"])
df.head()
```

## 12. Remove Extra Spaces from Text Columns

Text data mein extra spaces hidden hoti hain. Example: ` Ali ` ko `Ali` banana.

```python
df['student_name'] = df['student_name'].str.strip()
df['city'] = df['city'].str.strip()

df.head()
```

## 13. Standardize Text Case

City names aur student names ko proper format mein convert karte hain.

```python
df.head()
```

```python
df['student_name'] = df['student_name'].str.title()
df['city'] = df['city'].str.title()

df[['student_name', 'city']].head(10)
```

## 14. Check Unique Values

Unique values se spelling mistakes aur inconsistent values easily detect hoti hain.

```python
print("Cities:", df['city'].unique())
print("Gender:", df['gender'].unique())
```

## 15. Replace Wrong/Inconsistent Values

Examples:
- `Lhr` ko `Lahore`
- `Khi` ko `Karachi`
- `M` ko `Male`
- `F` ko `Female`

```python
df['city'] = df['city'].replace({
    "Lhr": "Lahore",
    "Khi": "Okara"
})

df['gender'] = df['gender'].replace({
    "M": "Male",
    "F": "Female"
})

print("Cities:", df['city'].unique())
print("Gender:", df['gender'].unique())
```

## 16. Handle Missing Values

Basic method:
- Numeric columns mein missing values ko mean/median se fill kar sakte hain
- Text columns mein missing values ko `Unknown` se fill kar sakte hain

```python
df.head()
```

```python
df['age'] = df['age'].fillna(df['age'].median())
df['marks'] = df['marks'].fillna(df['marks'].mean())
df['student_name'] = df['student_name'].fillna("Unknown")
df['city'] = df['city'].fillna("Unknown")
df['gender'] = df['gender'].fillna("Unknown")

df.isnull().sum()
```

## 17. Convert Date Column

Date column ko proper datetime format mein convert karte hain. Invalid dates ko `NaT` banaya jata hai using `errors='coerce'`.

```python
df["admission_date"].head()
```

```python
df['admission_date'] = pd.to_datetime(df['admission_date'], errors='coerce')

df[['admission_date']].head(15)
```

```python
df['admission_date'] = df['admission_date'].fillna(pd.Timestamp('2024-02-01'))
```

```python
df.isnull().sum()
```

## 18. Fill Missing Dates

Basic class ke liye missing date ko mode se fill kar dete hain.

```python
most_common_date = df['admission_date'].mode()[0]

print("Most common admission date:", most_common_date)



df['admission_date'] = df['admission_date'].fillna(most_common_date)

df.isnull().sum()
```

## 19. Remove Duplicate Rows

Ab duplicate rows remove karte hain.

```python
before = df.shape[0]
df = df.drop_duplicates()
after = df.shape[0]

print("Rows before removing duplicates:", before)
print("Rows after removing duplicates:", after)
```

## 20. Find Invalid Values

Age 0 se kam ya 100 se zyada ho to invalid maan sakte hain. Marks 0 se kam ya 100 se zyada ho to invalid hain.

```python
invalid_age = df[(df['age'] < 0) | (df['age'] > 100)]
invalid_marks = df[(df['marks'] < 0) | (df['marks'] > 100)]

print("Invalid Age Rows:")
display(invalid_age)

print("Invalid Marks Rows:")
display(invalid_marks)
```

```python
df[(df['age'] >= 18) & (df['marks'] >= 50)]
```

```python
df[~(df['city'] == 'Lahore')]
```

## 21. Remove Invalid Rows

Invalid age aur marks wali rows remove karte hain.

```python
df = df[(df['age'] >= 0) & (df['age'] <= 100)]
df = df[(df['marks'] >= 0) & (df['marks'] <= 100)]

df
```

## 22. Convert Data Types

Age ko integer mein convert karte hain. Marks ko 2 decimal tak round karte hain.

```python
df.head()
```

```python
df['age'] = df['age'].astype(int)
df['marks'] = df['marks'].round(2)

df.dtypes
```

## 23. Final Cleaned Dataset

Ab final cleaned dataset ready hai.

```python
df.reset_index(drop=True, inplace=True)
df
```

## 24. Save Cleaned Dataset

Cleaned dataset ko CSV file mein save kar lete hain.

```python
df.to_csv("cleaned_student_dataset.csv", index=False)
print("Cleaned dataset saved successfully!")
```
