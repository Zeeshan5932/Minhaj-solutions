# Topic 2: Normalization vs Standardization

There are two common scaling methods:

1. Normalization
2. Standardization

## 2.1 Normalization

Normalization converts values into a fixed range, usually between **0 and 1**.

### Formula

```text
X_normalized = (X - min) / (max - min)
```

### Simple Explanation

If a value is minimum, it becomes 0.

If a value is maximum, it becomes 1.

Other values come between 0 and 1.

### When to Use Normalization

Use normalization when:

- You want all values between 0 and 1
- You are using KNN
- You are using neural networks
- Data does not follow normal distribution

````python
minmax_scaler = MinMaxScaler()

df[["age_normalized", "fare_normalized"]] = minmax_scaler.fit_transform(df[["Age", "Fare"]])

print("Normalized Values:")
display(df[["Age", "age_normalized", "Fare", "fare_normalized"]])
````

## 2.2 Standardization

Standardization converts data around mean 0 and standard deviation 1.

### Formula

```text
X_standardized = (X - mean) / standard deviation
```

### Simple Explanation

After standardization:

- Mean becomes close to 0
- Standard deviation becomes close to 1

### When to Use Standardization

Use standardization when:

- You are using Logistic Regression
- You are using Linear Regression
- You are using SVM
- Data has outliers or different ranges

`11.18` **standard deviation** hai.

Humne data liya tha:

```text
Age = 20, 30, 40, 50
```

## Step 1: Mean nikaalo

```text
Mean = (20 + 30 + 40 + 50) / 4
Mean = 140 / 4
Mean = 35
```

## Step 2: Har value se mean minus karo

```text
20 - 35 = -15
30 - 35 = -5
40 - 35 = 5
50 - 35 = 15
```

## Step 3: In values ka square karo

```text
(-15)^2 = 225
(-5)^2  = 25
(5)^2   = 25
(15)^2  = 225
```

## Step 4: Squares ka average lo

```text
(225 + 25 + 25 + 225) / 4
= 500 / 4
= 125
```

## Step 5: Square root lo

```text
√125 = 11.18
```

So:

```text
Standard Deviation = 11.18
```




````python
standard_scaler = StandardScaler()

df[["age_standardized", "fare_standardized"]] = standard_scaler.fit_transform(df[["Age", "Fare"]])

print("Standardized Values:")
display(df[["Age", "age_standardized", "Fare", "fare_standardized"]])
````

## 2.3 Compare Normalization and Standardization

| Method | Output | Best Used For |
|---|---|---|
| Normalization | Values between 0 and 1 | KNN, Neural Networks |
| Standardization | Mean 0 and standard deviation 1 | SVM, Logistic Regression, Linear Regression |

### Important Point

Both methods are used for numerical data only.

We do not apply scaling directly on text columns.

````python
comparison = df[[
    "Age", "age_normalized", "age_standardized",
    "Fare", "fare_normalized", "fare_standardized"
]]

display(comparison)
````

# Topic 3: Encoding Categorical Data

## What is Categorical Data?

Categorical data means text-based data or group-based data.

Examples:

- Male, Female
- Lahore, Karachi, Islamabad
- First, Second, Third

Machine learning models cannot understand text directly.

So we convert text values into numbers. This process is called **encoding**.

````python
df.head()
````

````python
categorical_columns = ["Sex", "Embarked", "Pclass"]

display(df[categorical_columns])
````

## 3.1 Label Encoding

Label Encoding converts categories into numbers.

Example:

```text
male = 1
female = 0
```

### When to Use Label Encoding

Use label encoding when:

- Column has only two categories
- Categories have order, like Low, Medium, High

Here we will apply Label Encoding on `sex`.

````python
label_encoder = LabelEncoder()

df["sex_encoded"] = label_encoder.fit_transform(df["Sex"])

print("Label Encoding for Sex Column:")
display(df[["Sex", "sex_encoded"]])
````

## 3.2 One-Hot Encoding

One-Hot Encoding creates separate columns for each category.

Example:

If the column is `class`:

```text
First
Second
Third
```

After one-hot encoding, it becomes:

```text
class_First  class_Second  class_Third
1            0             0
0            1             0
0            0             1
```

### When to Use One-Hot Encoding

Use one-hot encoding when categories have no natural order.

Examples:

- City
- Country
- Embarked port
- Class name

````python
df.head()
````

````python
df_encoded = pd.get_dummies(df, columns=["Embarked", "Pclass"], drop_first=False)

print("Dataset After One-Hot Encoding:")
display(df_encoded)
````

## Label Encoding vs One-Hot Encoding

**Label Encoding** text categories ko numbers me convert karta hai, jaise `Low = 0`, `Medium = 1`, `High = 2`.  
Ye tab use hota hai jab categories me order/ranking ho, example: product size `Small < Medium < Large`.  

**One-Hot Encoding** har category ka separate column banata hai, jaise `City_Lahore`, `City_Karachi`, `City_Islamabad`.  
Ye tab use hota hai jab categories me koi order na ho, example: city, gender, country.  

ML me encoding is liye use hoti hai ky models text ko directly understand nahi karte, unko numbers chahiye hoty hain.  

Real example: `Education Level` ke liye Label Encoding use kar sakty hain, lekin `City Name` ke liye One-Hot Encoding better hai.
