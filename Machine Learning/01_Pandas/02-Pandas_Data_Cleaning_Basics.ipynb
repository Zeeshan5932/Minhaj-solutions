## Today’s Topic: Data Cleaning with Pandas

### 1. What is Data Cleaning?

Explain to students:

Data cleaning means fixing messy data before analysis or machine learning.

Real data usually has:

- Missing values
- Duplicate rows
- Wrong data types
- Extra spaces
- Incorrect column names
- Inconsistent text values
- Unwanted columns
- Outliers or wrong entries

Simple line for students:

**“Before we analyze data, we must clean it so the result becomes accurate.”**

# Complete Topics to Cover Today

## 1. Load Dataset

Start with importing pandas and reading a CSV file.

````python
import pandas as pd

df = pd.read_csv("data.csv")
df.head()
````

Explain:

- `pd.read_csv()` loads data
- `df.head()` shows first 5 rows

## 2. Check Dataset Shape

````python
df.shape
````

Explain:

This shows:

```text
(rows, columns)
```

Example:

```text
(1000, 8)
```

Means dataset has 1000 rows and 8 columns.

## 3. Check Column Names

````python
df.columns
````

Explain:

This shows all column names in the dataset.

## 4. Basic Dataset Information

````python
df.info()
````

Explain:

This tells us:

- Column names
- Non-null values
- Data types
- Total rows

Very important for first inspection.

## 5. Summary Statistics

````python
df.describe()
````

Explain:

This gives numerical summary:

- Count
- Mean
- Standard deviation
- Minimum
- Maximum

## 6. Check Missing Values

````python
df.isnull().sum()
````

Explain:

This shows how many missing values are present in each column.

Example:

```text
Name       0
Age        5
Salary     2
City       1
```

Means Age has 5 missing values.

## 7. Handle Missing Values

### Option 1: Remove Missing Rows

````python
df.dropna()
````

Explain:

This removes rows where any value is missing.

Use when missing data is very small.

### Option 2: Fill Missing Values

For numeric columns:

````python
df['Age'] = df['Age'].fillna(df['Age'].mean())
````

For text columns:

````python
df['City'] = df['City'].fillna("Unknown")
````

Explain:

- Mean is useful for numeric data
- “Unknown” is useful for missing text values

## 8. Check Duplicate Rows

````python
df.duplicated().sum()
````

Explain:

This tells how many duplicate rows exist.

## 9. Remove Duplicate Rows

````python
df = df.drop_duplicates()
````

Explain:

Duplicate data can affect analysis, so we remove it.

## 10. Rename Columns

Sometimes column names are messy:

```text
Student Name
student-age
Total Marks
```

Clean them like this:

````python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_")
````

Result:

```text
student_name
student-age
total_marks
```

For specific column rename:

````python
df = df.rename(columns={"student-age": "student_age"})
````

## 11. Remove Extra Spaces from Text

````python
df['Name'] = df['Name'].str.strip()
````

Explain:

Sometimes data has hidden spaces:

```text
" Ali "
```

After cleaning:

```text
"Ali"
```

## 12. Convert Text Case

````python
df['City'] = df['City'].str.lower()
````

or

````python
df['City'] = df['City'].str.title()
````

Explain:

This fixes inconsistent values like:

```text
lahore
Lahore
LAHORE
```

into:

```text
Lahore
```

## 13. Check Unique Values

````python
df['City'].unique()
````

Explain:

This shows all different values in one column.

Useful for finding spelling mistakes or inconsistent values.

## 14. Replace Wrong Values

Example:

````python
df['City'] = df['City'].replace("Lhr", "Lahore")
````

Multiple replacements:

````python
df['Gender'] = df['Gender'].replace({
    "M": "Male",
    "F": "Female"
})
````

## 15. Change Data Types

Check data types:

````python
df.dtypes
````

Convert column type:

````python
df['Age'] = df['Age'].astype(int)
````

Convert date column:

````python
df['Date'] = pd.to_datetime(df['Date'])
````

Explain:

Correct data type is important before analysis.

## 16. Drop Unnecessary Columns

````python
df = df.drop(columns=['Unnamed: 0'])
````

Explain:

Sometimes datasets contain useless columns. We remove them.

## 17. Filter Wrong Data

Example:

````python
df[df['Age'] < 0]
````

This checks invalid age values.

Remove wrong data:

````python
df = df[df['Age'] > 0]
````

## 18. Save Cleaned Dataset

````python
df.to_csv("cleaned_data.csv", index=False)
````

Explain:

After cleaning, save the clean file for future use.
