# Pandas Tutorial: Complete Guide (File 01_Pandas)

Pandas is a fast, powerful, and easy-to-use open-source data analysis and manipulation library built on top of Python. It is widely used in Machine Learning and Data Science workflows.

**Why use Pandas?**
- Efficient data manipulation and analysis.
- Handles structured data easily (CSV, JSON, Excel, SQL, etc.).
- Essential for data preprocessing in ML workflows.

**Importance in ML:**
- Data cleaning: handle missing values and duplicates.
- Feature engineering: create new features from existing data.
- Data exploration: summarize and visualize datasets quickly.


```
import pandas as pd
print('Pandas Version:', pd.__version__)
```

```
# Creating Series
series_from_list = pd.Series([10, 20, 30, 40, 50], name='Numbers')
series_from_dict = pd.Series({'a':100,'b':200,'c':300}, name='DictSeries')
print(series_from_list)
print(series_from_dict)
```

```
# Creating DataFrames
data = {
    'Name': ['Alice','Bob','Charlie','David'],
    'Age': [25,30,35,40],
    'City': ['New York','Los Angeles','Chicago','Houston']
}
df = pd.DataFrame(data)
print(df)
```

```
# Read CSV & JSON
# df_csv = pd.read_csv('data.csv')  # replace with actual file path
# df_json = pd.read_json('data.json')  # replace with actual file path
```

```
# Data Exploration
print(df.info())
print(df.describe())
print(df.head(3))
print(df.tail(2))
```

```
# Indexing & Filtering
print(df['Name'])
print(df[['Name','Age']])
print(df.iloc[1])
df_indexed = df.set_index('Name')
print(df_indexed.loc['Alice'])
print(df[df['Age'] > 30])
```

```
# Modify Columns & Handling Missing Data
df['Salary'] = [50000,60000,70000,80000]
df['Age'] = df['Age'] + 1
print(df)

df_missing = pd.DataFrame({'A':[1,2,None],'B':[None,5,6]})
print(df_missing.fillna(0))
print(df_missing.dropna())
```

```
# Sorting, Grouping & Merging
print(df.sort_values(by='Age'))
print(df.sort_values(by='Name', ascending=False))

df_group = pd.DataFrame({'Department':['HR','IT','HR','IT'],'Salary':[50000,60000,55000,65000]})
print(df_group.groupby('Department')['Salary'].mean())

df1 = pd.DataFrame({'ID':[1,2,3],'Name':['A','B','C']})
df2 = pd.DataFrame({'ID':[2,3,4],'Score':[90,95,85]})
merged_df = pd.merge(df1,df2,on='ID',how='inner')
print(merged_df)
```

```
# Saving Data
df.to_csv('output.csv', index=False)
df.to_json('output.json', orient='records')
```
