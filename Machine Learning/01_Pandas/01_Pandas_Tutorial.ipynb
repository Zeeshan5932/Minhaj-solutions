### Understanding `df.info()` in Pandas

The `df.info()` method in Pandas provides a concise summary of a DataFrame, which helps us quickly understand the structure and content of the dataset.


**What `df.info()` tells us:**

1. **Type of object**: `<class 'pandas.core.frame.DataFrame'>` shows that this is a Pandas DataFrame.
2. **Index range**: `RangeIndex: 5 entries, 0 to 4` indicates 5 rows with indices from 0 to 4.
3. **Columns info**:

   * Column names (`StudentID`, `Name`, `Age`, `Grade`, `City`)
   * Number of non-null entries in each column (`5 non-null` means no missing values)
   * Data type of each column (`int64` for numbers, `object` for text)
4. **Data types summary**: `dtypes: int64(2), object(3)` shows there are 2 numeric columns and 3 object (text) columns.
5. **Memory usage**: `memory usage: 332.0+ bytes` shows the memory consumed by the DataFrame.

**Why we use it:**

* Quickly check the structure of the dataset.
* Identify missing values.
* Understand data types for analysis or preprocessing.
* Optimize memory usage if working with large datasets.

---

````python
import pandas as pd

s = pd.Series([30, 10, 40, 20])
print("Original Series:\n", s)

# Sort ascending
s_sorted = s.sort_values()
print("\nSorted Series (Ascending):\n", s_sorted)

# Sort descending
s_sorted_desc = s.sort_values(ascending=False)
print("\nSorted Series (Descending):\n", s_sorted_desc)
````

````python
data = {'Name': ['Ali', 'Sara', 'Ahmed'], 'Age':[20, 22, 19], 'Score':[85, 90, 80]}
df = pd.DataFrame(data)
print("Original DataFrame:\n", df)

# Sort by single column
df_sorted_age = df.sort_values(by='Age')
print("\nSorted by Age:\n", df_sorted_age)

# Sort by multiple columns
df_sorted_age_score = df.sort_values(by=['Age', 'Score'], ascending=[False, True])
print("\nSorted by Age ascending, Score descending:\n", df_sorted_age_score)
````

## 6. Groupby & Aggregation

`groupby()` data ko groups mein split karta hai aur phir har group par summary calculation karta hai.
Aksar use hone wale aggregations:
- mean
- sum
- count
- min / max

````python
df_group = pd.DataFrame({
    'Department': ['HR', 'IT', 'HR', 'IT'],
    'Salary': [50000, 60000, 55000, 65000],
    'Employee': ['A', 'B', 'C', 'D']
})
df_group.head()


````

````python
print(df_group.groupby('Department')['Salary'].mean())
print(df_group.groupby('Department').agg(average_salary=('Salary', 'mean'), employee_count=('Employee', 'count')))
````

## 7. Merging & Joining DataFrames

Multiple DataFrames ko combine karne ke liye `merge()` use hota hai.
Common join types:
- `inner`
- `left`
- `right`
- `outer`

Join type choose karne se output rows ka behavior control hota hai.

````python
df1 = pd.DataFrame({'ID': [1, 2, 3], 'Name': ['A', 'B', 'C']})
df2 = pd.DataFrame({'ID': [2, 3, 4], 'Score': [90, 95, 85]})

````

````python
df1.head()
````

````python
df2.head()
````

````python
merged_inner = pd.merge(df1, df2, on='ID', how='inner')
merged_left = pd.merge(df1, df2, on='ID', how='left')

print('Inner join:')
print(merged_inner)
````

````python
print('\nLeft join:')
print(merged_left)
````
