## Cell 1: Import Required Libraries

First, we import the required Python libraries.

- `pandas` is used for data handling.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# This setting makes charts look cleaner
sns.set_theme(style="whitegrid")

# This command helps charts display inside Jupyter Notebook
%matplotlib inline
```

## Cell 2: Load Titanic Dataset Using Seaborn

Seaborn has some built-in practice datasets. Here we load the Titanic dataset.

This dataset contains passenger information such as age, gender, ticket class, fare, and survival status.

```python
df = sns.load_dataset('titanic')

# Display first 5 rows
df.head()
```

## Cell 3: Basic Dataset Information

Before creating graphs, we should understand the dataset.

- `shape` tells rows and columns.
- `info()` shows column names, missing values, and data types.
- `isnull().sum()` shows missing values in each column.

```python
print("Dataset Shape:", df.shape)
print("
Dataset Info:")
df.info()

print("
Missing Values:")
print(df.isnull().sum())
```

## Cell 4: Simple Data Cleaning for Visualization

For visualization, we handle a few missing values:

- Missing `age` values are filled with the median age.
- Missing `embarked` values are filled with the most common value.
- Missing `deck` values are filled with `Unknown`.

This step makes the dataset easier to visualize.

```python
df_clean = df.copy()

df_clean['age'] = df_clean['age'].fillna(df_clean['age'].median())
df_clean['embarked'] = df_clean['embarked'].fillna(df_clean['embarked'].mode()[0])
df_clean['deck'] = df_clean['deck'].cat.add_categories('Unknown').fillna('Unknown')

print("Missing values after basic cleaning:")
print(df_clean.isnull().sum())
```

## Graph 1: Line Plot - Passenger Index vs Fare

A line plot connects data points with a line. Here we use it to see how passenger fares change across passenger index numbers. This is mainly for learning line plot syntax.

### Student Explanation
Run the code and observe the chart. Ask students what the x-axis shows, what the y-axis shows, and what pattern they can see.

```python
plt.figure(figsize=(10,5))
plt.plot(df_clean.index[:100], df_clean['fare'].head(100), marker='o')
plt.title('Line Plot: First 100 Passenger Fares')
plt.xlabel('Passenger Index')
plt.ylabel('Fare')
plt.show()
```

## Graph 2: Bar Chart - Average Fare by Class

A bar chart compares values between categories. Here we compare the average fare paid by passengers in First, Second, and Third class.

### Student Explanation
Run the code and observe the chart. Ask students what the x-axis shows, what the y-axis shows, and what pattern they can see.

```python
avg_fare_class = df_clean.groupby('class')['fare'].mean()

plt.figure(figsize=(8,5))
plt.bar(avg_fare_class.index.astype(str), avg_fare_class.values)
plt.title('Average Fare by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Average Fare')
plt.show()
```

## Graph 21: Stacked Bar Chart - Class and Survival

A stacked bar chart shows how one category is divided by another category. Here we compare survival status inside each passenger class.

### Student Explanation
Run the code and observe the chart. Ask students what the x-axis shows, what the y-axis shows, and what pattern they can see.

```python
class_survival = pd.crosstab(df_clean['class'], df_clean['survived'])

class_survival.plot(kind='bar', stacked=True, figsize=(8,5))
plt.title('Stacked Bar Chart: Class vs Survival')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

## Graph 22: Area Plot - Cumulative Fare of First 50 Passengers

An area plot shows filled values over an index. Here we visualize cumulative fare for the first 50 passengers.

### Student Explanation
Run the code and observe the chart. Ask students what the x-axis shows, what the y-axis shows, and what pattern they can see.

```python
first_50 = df_clean.head(50).copy()
first_50['cumulative_fare'] = first_50['fare'].cumsum()

plt.figure(figsize=(10,5))
plt.fill_between(first_50.index, first_50['cumulative_fare'])
plt.title('Area Plot: Cumulative Fare of First 50 Passengers')
plt.xlabel('Passenger Index')
plt.ylabel('Cumulative Fare')
plt.show()
```

## Graph 28: Line Plot - Average Fare by Age Group

This graph first creates age groups, then shows average fare for each group. It helps students learn how grouping and visualization work together.

### Student Explanation
Run the code and observe the chart. Ask students what the x-axis shows, what the y-axis shows, and what pattern they can see.

```python
df_clean['age_group'] = pd.cut(df_clean['age'], bins=[0, 12, 18, 30, 45, 60, 100], labels=['Child', 'Teen', 'Young Adult', 'Adult', 'Middle Age', 'Senior'])
age_fare = df_clean.groupby('age_group', observed=True)['fare'].mean().reset_index()

plt.figure(figsize=(10,5))
sns.lineplot(data=age_fare, x='age_group', y='fare', marker='o')
plt.title('Average Fare by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Average Fare')
plt.xticks(rotation=30)
plt.show()
```
