## Matplotlib & Seaborn Charts in Titanic Notebook

Here is a summary of the charts used in the notebook, categorized by **categorical** and **numerical** usage.

| Chart Type   | Columns Used                                     | Data Type               | Purpose                                                                   |
| ------------ | ------------------------------------------------ | ----------------------- | ------------------------------------------------------------------------- |
| Count Plot   | `Sex`, `Pclass`, `Embarked`, `Survived`          | Categorical             | Shows counts/frequency of categories                                      |
| Bar Plot     | `Pclass` vs `Survived`, `Embarked` vs `Survived` | Categorical             | Compare category values across another categorical/numerical target       |
| Box Plot     | `Age` vs `Pclass`, `Fare` vs `Survived`          | Numerical               | Shows distribution, median, quartiles, outliers                           |
| Violin Plot  | `Age` vs `Pclass`, `Fare` vs `Sex`               | Numerical               | Similar to box plot but shows density distribution                        |
| Histogram    | `Age`, `Fare`                                    | Numerical               | Distribution of single numerical variable                                 |
| KDE Plot     | `Age`, `Fare`                                    | Numerical               | Smoothed density plot of a numerical variable                             |
| Scatter Plot | `Age` vs `Fare`                                  | Numerical               | Shows relationship between two numerical variables                        |
| Joint Plot   | `Age` vs `Fare`                                  | Numerical + Categorical | Combines scatter with histogram/KDE for numerical relationship            |
| Pair Plot    | `Age`, `Fare`, `Pclass`                          | Numerical + Categorical | Pairwise scatterplots for multiple numerical columns, hue for categorical |
| Swarm Plot   | `Age` vs `Pclass`                                | Numerical + Categorical | Shows all points distribution for categorical groups                      |
| Heatmap      | Correlation matrix                               | Numerical               | Shows correlation strength between numerical variables                    |
| Line Plot    | `Age` vs `Survived` counts (aggregated)          | Numerical               | Trends over ordered data                                                  |
| Area Plot    | `Survived` counts over `Pclass`                  | Categorical             | Cumulative distribution of numerical counts across categories             |
| Pie Chart    | `Survived`                                       | Categorical             | Shows proportion of categories                                            |
| LM Plot      | `Age` vs `Fare`, hue=`Sex`                       | Numerical + Categorical | Regression line between numerical variables with category separation      |
| Facet Grid   | `Age` vs `Fare` split by `Sex`                   | Numerical + Categorical | Multiple plots for each category                                          |

**Notes for Students:**

* **Categorical charts** are mainly: Count Plot, Bar Plot, Pie Chart, Facet Grid (for categories).
* **Numerical charts** are mainly: Histogram, KDE, Box Plot, Violin Plot, Scatter, Joint, Pair, Swarm, Line, Area, LM Plot, Heatmap.
* Some charts combine both types, e.g., `Box Plot` with numerical on y-axis and categorical on x-axis.
* Always choose chart type based on the **data type** of your column(s).

## Cell 4: Simple Data Cleaning for Visualization

For visualization, we handle a few missing values:

- Missing `age` values are filled with the median age.
- Missing `embarked` values are filled with the most common value.
- Missing `deck` values are filled with `Unknown`.

This step makes the dataset easier to visualize.

```python
df_clean = df.copy()

df_clean['age'] = df_clean['age'].fillna(df_clean['age'].median())
df_clean['embarked'] = df_clean['embarked'].fillna(df_clean['embarked'].mode()[0])
df_clean['deck'] = df_clean['deck'].cat.add_categories('Unknown').fillna('Unknown')

print("Missing values after basic cleaning:")
print(df_clean.isnull().sum())
```
